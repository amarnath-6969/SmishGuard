```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set visual style
sns.set_theme(style="whitegrid")

print("--- 1. LOADING DATA ---")
# Load Seaborn's built-in Diamonds dataset (53,940 rows, 10 columns)
df = sns.load_dataset('diamonds')
print("Dataset loaded successfully!\n")

print("--- 2. STRUCTURAL OVERVIEW ---")
print(f"Shape: {df.shape[0]} rows, {df.shape[1]} columns")
print("-" * 40)
print("Data Types & Memory:")
df.info()
print("-" * 40)
print("First 5 Rows:")
display(df.head())
print("\nLast 5 Rows:")
display(df.tail())
print("\n" + "="*50 + "\n")

print("--- 3. MISSING VALUE ANALYSIS ---")
missing = df.isnull().sum()
print("Missing values per column:\n", missing)

plt.figure(figsize=(8, 4))
sns.heatmap(df.isnull(), cbar=False, cmap='viridis', yticklabels=False)
plt.title("Missing Value Heatmap", fontsize=12, fontweight='bold')
plt.show()
print("\n" + "="*50 + "\n")

print("--- 4. SUMMARY STATISTICS & VALUE COUNTS ---")
print("Numeric Summary Statistics:")
display(df.describe())

print("\nValue Counts for 'cut':")
display(df['cut'].value_counts())

print("\nValue Counts for 'color':")
display(df['color'].value_counts())
print("\n" + "="*50 + "\n")

print("--- 5. GENERATING 5 VISUALIZATIONS ---")

# Visualization 1: Histogram & KDE (Price Distribution)
plt.figure(figsize=(9, 4))
sns.histplot(df['price'], kde=True, color='purple', bins=50)
plt.title("Distribution of Diamond Prices", fontsize=12, fontweight='bold')
plt.xlabel("Price (USD)")
plt.ylabel("Frequency")
plt.show()
print("Interpretation: Price distribution is heavily right-skewed; most diamonds cost under $5,000.\n")

# Visualization 2: Correlation Heatmap
plt.figure(figsize=(7, 5))
numeric_df = df.select_dtypes(include=[np.number])
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Correlation Heatmap", fontsize=12, fontweight='bold')
plt.show()
print("Interpretation: Carat weight and physical dimensions (x, y, z) show a strong positive correlation (>0.88) with price.\n")

# Visualization 3: Scatter Plot (Price vs. Carat by Clarity)
plt.figure(figsize=(9, 5))
sns.scatterplot(data=df, x='carat', y='price', alpha=0.3, hue='clarity', palette='viridis')
plt.title("Price vs. Carat Weight by Clarity Grade", fontsize=12, fontweight='bold')
plt.xlabel("Carat")
plt.ylabel("Price (USD)")
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
print("Interpretation: Non-linear exponential trend between carat and price; lower clarity grades sit lower on the pricing curve.\n")

# Visualization 4: Box Plot (Price across Cut Grades)
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='cut', y='price', palette='Set2')
plt.title("Price Distribution across Cut Quality", fontsize=12, fontweight='bold')
plt.xlabel("Cut Quality")
plt.ylabel("Price (USD)")
plt.show()
print("Interpretation: 'Ideal' cuts often show lower median prices than lower-tier cuts, as buyers trade off cut grade for larger stones.\n")

# Visualization 5: Bar Plot (Diamond Count by Color & Cut)
plt.figure(figsize=(10, 5))
sns.countplot(data=df, x='color', hue='cut', palette='muted')
plt.title("Diamond Count by Color and Cut Grade", fontsize=12, fontweight='bold')
plt.xlabel("Color Grade")
plt.ylabel("Count")
plt.legend(title="Cut")
plt.show()
print("Interpretation: Color grades G and E are the most abundant, and 'Ideal' cuts dominate most color categories.\n")

print("="*50)
print("--- 6. WRITTEN SUMMARY OF KEY FINDINGS ---")
print("""
1. Primary Cost Driver: Carat weight and physical dimensions (x, y, z) show the highest positive correlation (>0.88) with price[span_5](start_span)[span_5](end_span).
2. Skewed Price Distribution: Diamond prices cluster heavily under $5,000, with a sparse long tail going up to ~$19,000[span_6](start_span)[span_6](end_span).
3. Quality vs. Carat Trade-off: Lower cut quality grades often feature higher median prices, indicating that buyers trade off cut precision to buy larger carat weights[span_7](start_span)[span_7](end_span).
4. Clean Dataset: The dataset contains zero missing values across all 53,940 records[span_8](start_span)[span_8](end_span).
""")
```
